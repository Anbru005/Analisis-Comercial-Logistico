### 3.3 Convierte las fechas a formato datetime y crea nuevas columnas

```python
# 2. CONVERSIÓN DE FECHAS
df_consolidado['Fecha Facturación'] = pd.to_datetime(df_consolidado['Fecha Facturación'])

# 3. CREACIÓN DE NUEVAS COLUMNAS (VARIABLES)
df_consolidado['Año'] = df_consolidado['Fecha Facturación'].dt.year
df_consolidado['Mes'] = df_consolidado['Fecha Facturación'].dt.month

# Calculamos el margen según tu fórmula (Venta - Impuesto)
df_consolidado['Margen'] = df_consolidado['Monto Total'] - df_consolidado['Impuesto']

# Calculamos el ratio de impuestos protegiendo el código si hay ventas en cero
df_consolidado['Ratio de impuestos'] = df_consolidado['Impuesto'] / df_consolidado['Monto Total'].replace(0, np.nan)

# NOTA: 'Días de entrega' ya lo calculamos eficientemente en la vista de SQL.
# Aquí solo nos aseguramos de que Python lo guarde formalmente con el nombre exacto solicitado.
df_consolidado['Días de entrega'] = df_consolidado['Días de Entrega']

# 4. VALIDACIÓN VISUAL
print("📌 MUESTRA DE LAS NUEVAS COLUMNAS GENERADAS:")
columnas_reporte = ['Fecha Facturación', 'Año', 'Mes', 'Monto Total', 'Impuesto', 'Margen', 'Ratio de impuestos', 'Días de entrega']
print(df_consolidado[columnas_reporte].head())
```

### 3.4 Calcula el total de ventas, promedio, mediana y desviación estándar por país.

```python
# 1. Agrupar por país y calcular las métricas
estadisticas_pais = df_consolidado.groupby('País')['Monto Total'].agg(
    Total_Ventas='sum',           # Suma total
    Promedio='mean',              # Promedio aritmético
    Mediana='median',             # Valor central
    Desviacion_Estandar='std'     # Dispersión de los datos
).reset_index()

# 2. Redondear los resultados a 2 decimales para formato de moneda
estadisticas_pais = estadisticas_pais.round(2)

# 3. Mostrar el resultado ordenado por el total de ventas (de mayor a menor)
estadisticas_pais = estadisticas_pais.sort_values(by='Total_Ventas', ascending=False)

print("📌 ESTADÍSTICAS FINANCIERAS POR PAÍS:")
print("-" * 65)
display(estadisticas_pais)
```

### 3.5 Detecta los registros fuera del rango normal (outliers)

```python
# 1. EL DIAGNÓSTICO INICIAL CON DESCRIBE()
print("📌 RESUMEN ESTADÍSTICO (MONTO TOTAL):")
# Guardamos el describe en una variable para extraer los datos
resumen_monto = df_consolidado['Monto Total'].describe()
print(resumen_monto.round(2))
print("-" * 50)

# 2. CÁLCULO DE LOS LÍMITES MATEMÁTICOS (Fórmula IQR)
# Extraemos el Cuartil 1 (25%) y el Cuartil 3 (75%)
Q1 = resumen_monto['25%']
Q3 = resumen_monto['75%']

# Calculamos el Rango Intercuartílico (La caja central donde está la mayoría)
IQR = Q3 - Q1

# Definimos las "fronteras" de normalidad (El multiplicador 1.5 es el estándar estadístico)
limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

print("📌 FRONTERAS DE NORMALIDAD:")
print(f"Límite Inferior: ${limite_inferior:,.2f}")
print(f"Límite Superior: ${limite_superior:,.2f}")
print("-" * 50)

# 3. ATRAPAR LOS REGISTROS FUERA DE RANGO
# Filtramos la tabla buscando lo que sea menor al límite inferior O mayor al superior
outliers = df_consolidado[(df_consolidado['Monto Total'] < limite_inferior) | 
                        (df_consolidado['Monto Total'] > limite_superior)]

# 4. REPORTE FINAL
cantidad_outliers = len(outliers)
porcentaje_outliers = (cantidad_outliers / len(df_consolidado)) * 100

print(f"🚨 ALERTA DE OUTLIERS:")
print(f"Se detectaron {cantidad_outliers} registros atípicos (equivale al {porcentaje_outliers:.2f}% de tu base de datos).")
print("\n📌 Muestra de las facturas atípicas más extremas:")

# Mostramos los outliers ordenados de mayor a menor para ver las compras más exageradas
display(outliers[['Fecha Facturación', 'Cliente', 'País', 'Monto Total']].sort_values(by='Monto Total', ascending=False).head())
```

### 3.6 Clasifica los clientes según su nivel de facturación (alta, media, baja)

```python
# 1. CONSOLIDAR VENTAS POR CLIENTE
# Agrupamos por cliente y sumamos todo su historial de compras
facturacion_clientes = df_consolidado.groupby('Cliente').agg(
    Total_Comprado=('Monto Total', 'sum')
).reset_index()

# 2. CLASIFICACIÓN ESTADÍSTICA (qcut)
# Dividimos a los clientes en 3 grupos (q=3) y les asignamos una etiqueta
etiquetas = ['Baja', 'Media', 'Alta']
facturacion_clientes['Nivel Facturación'] = pd.qcut(facturacion_clientes['Total_Comprado'], q=3, labels=etiquetas)

# 3. REPORTE DE RESULTADOS (¿Cuántos clientes hay en cada segmento?)
resumen_segmentos = facturacion_clientes['Nivel Facturación'].value_counts().reset_index()
resumen_segmentos.columns = ['Segmento', 'Cantidad de Clientes']
# Ordenamos para que se vea lógico: Alta, Media, Baja
resumen_segmentos['Orden'] = [1, 2, 3] # Asumiendo el orden de value_counts, pero mejor ordenamos por la categoría nativa
resumen_segmentos = resumen_segmentos.sort_values(by='Segmento', ascending=False)

print("📌 RESUMEN DE LA CARTERA DE CLIENTES:")
print("-" * 45)
print(resumen_segmentos[['Segmento', 'Cantidad de Clientes']].to_string(index=False))

# 4. MUESTRA DEL TOP CLIENTES (Categoría VIP / Alta)
# Filtramos solo los de nivel 'Alta' y los ordenamos de mayor a menor
clientes_vip = facturacion_clientes[facturacion_clientes['Nivel Facturación'] == 'Alta']
clientes_vip = clientes_vip.sort_values(by='Total_Comprado', ascending=False)

print("\n📌 TOP 5 CLIENTES VIP (Categoría ALTA):")
print("-" * 45)
display(clientes_vip.head())
```

### 3.7 Identifica los productos con alta rentabilidad y bajo volumen de ventas

```python
import pandas as pd

# 1. CONSOLIDAR VOLUMEN Y RENTABILIDAD POR PRODUCTO
# Asumiendo que 'Margen (%)' ya venía en tu vista o lo calculaste como (Margen / Monto Total) * 100
analisis_productos = df_consolidado.groupby('Producto').agg(
    Volumen_Ventas=('Monto Total', 'sum'),
    Rentabilidad_Promedio=('Margen (%)', 'mean')
).reset_index()

# 2. CALCULAR LOS PUNTOS DE CORTE (MEDIANAS)
# La mediana divide el catálogo exactamente en 50% de productos altos y 50% bajos
mediana_volumen = analisis_productos['Volumen_Ventas'].median()
mediana_rentabilidad = analisis_productos['Rentabilidad_Promedio'].median()

print("📊 LÍMITES OPERATIVOS DE LOS CUADRANTES:")
print(f"Umbral de Volumen (Mediana de Ventas): ${mediana_volumen:,.2f}")
print(f"Umbral de Rentabilidad (Mediana de Margen): {mediana_rentabilidad:.2f}%\n")

# 3. FILTRAR EL CUADRANTE: ALTA RENTABILIDAD Y BAJO VOLUMEN
# Usamos el operador lógico '&' (AND) para cumplir ambas condiciones
productos_nicho = analisis_productos[
    (analisis_productos['Rentabilidad_Promedio'] >= mediana_rentabilidad) & 
    (analisis_productos['Volumen_Ventas'] < mediana_volumen)
]

# Ordenamos por rentabilidad para ver los productos más eficientes primero
productos_nicho = productos_nicho.sort_values(by='Rentabilidad_Promedio', ascending=False)

print(f"🚨 MATRIZ DE PORTAFOLIO: Se identificaron {len(productos_nicho)} productos en el cuadrante de Nicho.")
print("-" * 75)
display(productos_nicho.head(10))
```

### 3.8 Determina si existen diferencias significativas entre países o categorías de cliente

```python
import scipy.stats as stats

# ---------------------------------------------------------
# 1. EVALUACIÓN POR PAÍS (Con validación de seguridad)
# ---------------------------------------------------------
print("📌 PRUEBA DE HIPÓTESIS: VENTAS POR PAÍS")
montos_por_pais = [grupo['Monto Total'].values for nombre, grupo in df_consolidado.groupby('País')]

# Validamos que existan al menos 2 países en la base de datos
if len(montos_por_pais) >= 2:
    estadistico_p, p_valor_pais = stats.kruskal(*montos_por_pais)
    print(f"Valor P (p-value): {p_valor_pais}")
    if p_valor_pais < 0.05:
        print("✅ CONCLUSIÓN: SÍ existen diferencias estadísticamente significativas en las ventas entre países.")
    else:
        print("❌ CONCLUSIÓN: NO existen diferencias significativas.")
else:
    print("⚠️ ANÁLISIS CANCELADO: La base de datos solo contiene operaciones en un (1) país.")
    print("No es matemáticamente posible realizar una comparación de varianza.")

print("\n" + "-" * 50 + "\n")

# ---------------------------------------------------------
# 2. EVALUACIÓN POR CATEGORÍA DE CLIENTE (Alta, Media, Baja)
# ---------------------------------------------------------
print("📌 PRUEBA DE HIPÓTESIS: CATEGORÍAS DE CLIENTE")
# Extraemos los montos y limpiamos posibles grupos vacíos generados por pd.qcut
montos_por_categoria = [grupo['Total_Comprado'].values for nombre, grupo in facturacion_clientes.groupby('Nivel Facturación') if len(grupo) > 0]

# Validamos que existan al menos 2 categorías con datos
if len(montos_por_categoria) >= 2:
    estadistico_c, p_valor_categoria = stats.kruskal(*montos_por_categoria)
    print(f"Valor P (p-value): {p_valor_categoria}")
    
    if p_valor_categoria < 0.05:
        print("✅ CONCLUSIÓN: SÍ existen diferencias estadísticamente significativas en el volumen de compra entre los niveles Alta, Media y Baja.")
    else:
        print("❌ CONCLUSIÓN: NO existen diferencias significativas.")
else:
    print("⚠️ ANÁLISIS CANCELADO: No hay suficientes categorías con datos para comparar.")
```

### 4.7 Curva de Crecimiento Global Acumulado por Estado

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. IDENTIFICAR LOS TOP 3 ESTADOS MOTORES
top_3_estados = df_consolidado.groupby('Estado')['Monto Total'].sum().nlargest(3).index

# 2. FILTRAR Y ORDENAR CRONOLÓGICAMENTE
# Es obligatorio ordenar por Año y Mes antes de calcular un acumulado
df_crecimiento = df_consolidado[df_consolidado['Estado'].isin(top_3_estados)].copy()
df_crecimiento = df_crecimiento.sort_values(by=['Año', 'Mes'])

# 3. AGRUPAR FACTURACIÓN MENSUAL POR ESTADO
crecimiento_temporal = df_crecimiento.groupby(['Estado', 'Año', 'Mes'])['Monto Total'].sum().reset_index()

# 4. CREACIÓN DE LA NUEVA COLUMNA: Suma Acumulada (cumsum)
# Agrupamos por Estado para que el acumulado de Texas no se mezcle con el de California
crecimiento_temporal['Ventas_Acumuladas'] = crecimiento_temporal.groupby('Estado')['Monto Total'].cumsum()

# Creamos una etiqueta limpia para el eje X (Formato: Año-Mes)
crecimiento_temporal['Periodo'] = crecimiento_temporal['Año'].astype(str) + "-" + crecimiento_temporal['Mes'].astype(str).str.zfill(2)

# 5. DISEÑO DE LA VISUALIZACIÓN
plt.figure(figsize=(14, 6))
sns.set_theme(style="whitegrid")

ax = sns.lineplot(
    data=crecimiento_temporal,
    x='Periodo',
    y='Ventas_Acumuladas',
    hue='Estado',
    marker='o',
    linewidth=2.5,
    palette='Set1'
)

# 6. PERSONALIZACIÓN ESTÉTICA
plt.title('Curva de Crecimiento Global Acumulado por Estado', fontsize=14, pad=15)
plt.xlabel('Línea de Tiempo Cronológica (Año-Mes)', fontsize=12)
plt.ylabel('Inyección de Capital Acumulado ($)', fontsize=12)

# Rotamos las etiquetas del eje X para que no se encimen
plt.xticks(rotation=45, ha='right')

# Formato de moneda para el eje Y
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, loc: f"${x:,.0f}"))

plt.tight_layout()
plt.show()
```

### 4.8 Gráfico combinado de ventas vs margen promedio

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ---------------------------------------------------------
# 1. PREPARACIÓN DE LOS DATOS
# ---------------------------------------------------------
# Agrupamos sumando la Ganancia Neta y el Monto Total (Igual a los SUM de DAX)
analisis_mensual = df_consolidado.groupby('Mes').agg(
    Ventas_Totales=('Monto Total', 'sum'),
    Ganancia_Total=('Ganancia Neta', 'sum') 
).reset_index()

# Calculamos el porcentaje agregado al final (Igual al DIVIDE de DAX)
analisis_mensual['Margen_Promedio'] = (analisis_mensual['Ganancia_Total'] / analisis_mensual['Ventas_Totales']) * 100


# ---------------------------------------------------------
# 2. CONFIGURACIÓN DEL LIENZO
# ---------------------------------------------------------
sns.set_theme(style="white") # Fondo blanco limpio
fig, ax1 = plt.subplots(figsize=(12, 6))


# ---------------------------------------------------------
# 3. EJE PRINCIPAL (Izquierdo): Barras de Volumen de Ventas
# ---------------------------------------------------------
color_barras = 'steelblue'
ax1.bar(analisis_mensual['Mes'], analisis_mensual['Ventas_Totales'], color=color_barras, alpha=0.7)
ax1.set_xlabel('Mes del Año', fontsize=12)
ax1.set_ylabel('Facturación Total ($)', color=color_barras, fontsize=12, fontweight='bold')
ax1.tick_params(axis='y', labelcolor=color_barras)

# Formato de moneda, nombres de meses y límites del eje X
ax1.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, loc: f"${x:,.0f}"))
meses_nombres = ['Ene', 'Feb', 'Mar', 'Abr', 'May', 'Jun', 'Jul', 'Ago', 'Sep', 'Oct', 'Nov', 'Dic']
ax1.set_xticks(range(1, 13))
ax1.set_xticklabels(meses_nombres)


# ---------------------------------------------------------
# 4. EJE SECUNDARIO (Derecho): Línea de Margen Neto
# ---------------------------------------------------------
ax2 = ax1.twinx() # Clona el eje X creando un nuevo eje Y
color_linea = 'darkorange'
# Graficamos la línea de margen con los datos corregidos
ax2.plot(analisis_mensual['Mes'], analisis_mensual['Margen_Promedio'], color=color_linea, marker='o', linewidth=3.5, markersize=8)
ax2.set_ylabel('Margen de Ganancia Neta (%)', color=color_linea, fontsize=12, fontweight='bold')
ax2.tick_params(axis='y', labelcolor=color_linea)

# Para asegurar que la línea tenga un respiro visual y no se pegue a los bordes
# Obligamos a Python a iniciar el eje Y secundario en un valor un poco más bajo que el mínimo (ej. 40% a 50%)
min_margin = analisis_mensual['Margen_Promedio'].min() - 2
max_margin = analisis_mensual['Margen_Promedio'].max() + 2
ax2.set_ylim(min_margin, max_margin)


# ---------------------------------------------------------
# 5. TOQUES FINALES
# ---------------------------------------------------------
plt.title('Eficiencia Comercial: Volumen de Ventas vs. Margen Neto (43-44%)', fontsize=14, pad=15)
fig.tight_layout() # Evita que los textos se corten
plt.show()
```

### 4.11 Mapa con puntos proporcionales a las ventas

```python
import plotly.express as px
import pandas as pd

# 1. Agrupamos las ventas totales por Estado
ventas_estado = df_consolidado.groupby('Estado')['Monto Total'].sum().reset_index()

# 2. Diccionario auxiliar para convertir nombres a códigos (Requerido por los mapas de EE.UU.)
codigos_estados = {
    'Texas': 'TX', 'Pennsylvania': 'PA', 'California': 'CA', 'New York': 'NY',
    'Florida': 'FL', 'Ohio': 'OH', 'Minnesota': 'MN', 'New Jersey': 'NJ',
    'Missouri': 'MO', 'New Mexico': 'NM', 'Washington': 'WA', 'Illinois': 'IL',
    'Colorado': 'CO', 'Georgia': 'GA', 'North Carolina': 'NC', 'Virginia': 'VA'
}

# Aplicamos la traducción
ventas_estado['Codigo_Estado'] = ventas_estado['Estado'].map(codigos_estados)

# Filtramos los datos listos para graficar
ventas_mapa = ventas_estado.dropna(subset=['Codigo_Estado'])

# 3. Crear el mapa de burbujas proporcionales (Scatter Geo)
fig = px.scatter_geo(
    ventas_mapa, 
    locations="Codigo_Estado", 
    locationmode="USA-states", 
    scope="usa",                   # Enfocamos la cámara solo en Estados Unidos
    size="Monto Total",            # El tamaño del punto depende de las ventas
    color="Monto Total",           # El color también cambia según el volumen
    hover_name="Estado",           # Texto al pasar el mouse
    title="Concentración de Clientes VIP: Volumen de Facturación Territorial",
    color_continuous_scale="Plasma",
    size_max=50                    # Tamaño máximo de la burbuja más grande
)

# 4. Mostrar el mapa interactivo
fig.show()
```

### 4.12 Mapa Coroplético según el tiempo promedio de entrega

```python
import plotly.express as px
import pandas as pd

# 1. PREPARACIÓN DE DATOS
# Agrupamos por Estado y calculamos el Promedio (mean) de los días de entrega
logistica_promedio = df_consolidado.groupby('Estado')['Días de Entrega'].mean().reset_index()

# Redondeamos a un decimal para que el mapa se lea más limpio
logistica_promedio['Días de Entrega Promedio'] = logistica_promedio['Días de Entrega'].round(1)

# 2. TRADUCCIÓN GEOGRÁFICA
codigos_estados = {
    'Texas': 'TX', 'Pennsylvania': 'PA', 'California': 'CA', 'New York': 'NY',
    'Florida': 'FL', 'Ohio': 'OH', 'Minnesota': 'MN', 'New Jersey': 'NJ',
    'Missouri': 'MO', 'New Mexico': 'NM', 'Washington': 'WA', 'Illinois': 'IL',
    'Colorado': 'CO', 'Georgia': 'GA', 'North Carolina': 'NC', 'Virginia': 'VA'
}

logistica_promedio['Codigo_Estado'] = logistica_promedio['Estado'].map(codigos_estados)
mapa_datos = logistica_promedio.dropna(subset=['Codigo_Estado'])

# 3. DISEÑO DEL MAPA COROPLÉTICO
fig = px.choropleth(
    mapa_datos,
    locations='Codigo_Estado',
    locationmode="USA-states",
    color='Días de Entrega Promedio',
    scope="usa",
    color_continuous_scale="Reds", # Escala de rojos para resaltar el "peligro" o ineficiencia
    title="Auditoría de Cadena de Suministro: Tiempo Promedio de Entrega por Estado",
    hover_name="Estado",
    labels={'Días de Entrega Promedio': 'Días (Promedio)'}
)

# 4. Mostrar el mapa
fig.show()
```
